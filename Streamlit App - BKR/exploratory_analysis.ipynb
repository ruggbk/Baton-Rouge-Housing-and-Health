```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import folium
```

```python
housing_data = pd.read_csv(r'../Output Data/housing_data.csv')
```

## GeoJSON plot

```python
# Load GeoJSON
gdf = gpd.read_file(r'../Output Data/housing_data_with_geometry.geojson')

# Plot Median Household Income
fig, ax = plt.subplots(figsize=(10,8))
gdf.plot(column='Median_Household_Income', 
         cmap='OrRd', 
         linewidth=0.8, 
         edgecolor='black', 
         legend=True, 
         ax=ax)
ax.set_title("Median Household Income by Census Tract")
ax.axis('off')
plt.show()
```

```python
import folium

# Create a base map
m = folium.Map(location=[30.45, -91.15], zoom_start=12)

# Add a choropleth
folium.Choropleth(
    geo_data=gdf,
    data=gdf,
    columns=['GEOID', 'Median_Household_Income'],
    key_on='feature.properties.GEOID',
    fill_color='YlOrRd',
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name='Median Household Income'
).add_to(m)

# Display in Jupyter
m
```
